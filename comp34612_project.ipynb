## **Game Rules and Playing Scenarios**

**Game Rules**

On day $t$, the leader announces the price, $u_L (t)$, first. After learning the leader’s price, the follower will choose the responding price, $u_F (t)$. This process occurs over a period of $t=1,2,…,130$, where $t=1,2,…,100$ corresponds to the days for which historical data is available, and $t=101,102,…,130$ are the days during which the game will be played repeatedly.


**Game Playing Scenarios**

In this setup, your program, acting as the leader, will engage in three separate games against three different competitors, identified as followers MK1, MK2, and MK3.

For each of the repeated game, the leader selects his learning method to learn the follower’s reaction function based on the provided set of historical data which spans from $t=1,2,…,100$. During the next 30 days ($t=101,102,…,130$), the leader announces a price, $u_L (t)$, to the game platform daily. After the leader's price is announced, the follower will respond with the price, $u_F (t)$. Upon receiving the follower's price, $u_F (t)$, from the game platform, the leader may use this new information to refine their knowledge/ update the learned reaction function and decide his price of the next day. In the other words, the leader (i.e., your program) should be able to take the follower’s price of the previous day from the game platform for updating, and then decide and send his price of the next day to the game platform.

It should be emphasised that the follower’s strategy and payoff function is subject to the changing and time varying environment. That is, the parameters in the follower’s payoff function may not be the same every day.

## Base Leader

This is the base class for all leader subclasses.

```python
class Leader:
    _subclass_registry = {}

    def __init__(self, name, engine):
        self.name = name
        self.engine = engine

    @classmethod
    def cleanup_old_subclasses(cls):
        """
        A function to remove old subclasses before defining new ones.
        """
        existing_subclasses = list(cls.__subclasses__())

        for subclass in existing_subclasses:
            subclass_name = subclass.__name__
            if subclass_name in cls._subclass_registry:
                del cls._subclass_registry[subclass_name]
                del subclass
        gc.collect()

    @classmethod
    def update_subclass_registry(cls):
        """
        A function to update registry after cleaning up old subclasses.
        """
        cls.cleanup_old_subclasses()
        cls._subclass_registry = {subclass.__name__: subclass for subclass in cls.__subclasses__()}

    def new_price(self, date):
        """
        A function for setting the new price of each day.
        :param date: date of the day to be updated
        :return: (float) price for the day
        """
        pass

    def get_price_from_date(self, date):
        """
        A function for getting the price set on a date.
        :param date: (int) date to get the price from
        :return: a tuple (leader_price, follower_price)
        """
        return self.engine.exposed_get_price(date)


    def start_simulation(self):
        """
        A function runs at the beginning of the simulation.
        """
        pass

    def end_simulation(self):
        """
        A function runs at the beginning of the simulation.
        """
        pass
```

```python
# =============================================================================
# Implementation Approach: Multiple Leaders
# -----------------------------------------------------------------------------
# LeaderMK1  →  designed to play against MK1 (and similar hidden follower MK4)
# LeaderMK2  →  designed to play against MK2 (and similar hidden follower MK5)
# LeaderMK3  →  designed to play against MK3 (and similar hidden follower MK6)
#
# Strategy: Adaptive Stackelberg Pricing
# ----------------------------------------
# Each follower is assumed to have a linear reaction function:
#     u_F = a * u_L + b
# Plugging into the leader demand model  S_L = 100 - 5*u_L + 3*u_F:
#     S_L = (100 + 3b) + (3a - 5)*u_L  =  A + B*u_L
#
# Leader profit:  pi_L = (u_L - 1) * (A + B*u_L)
# Setting d(pi_L)/d(u_L) = 0 gives the Stackelberg optimal price:
#     u_L*  =  (B - A) / (2B)       [valid when B < 0, i.e. a < 5/3]
#
# NOTE: _EWLSMixin does NOT inherit from Leader so that LeaderMK1/2/3 can
# inherit directly from Leader — required because the GUI populates its
# dropdown using Leader.__subclasses__() which only returns direct subclasses.
# =============================================================================

class _RLSMixin:
    """Recursive Least Squares + Stackelberg pricing (Lecture 6).
    Must be combined with a Leader subclass."""

    def _init_rls(self, min_price=1.0, max_price=100.0, lam=0.97):
        self.min_price = min_price
        self.max_price = max_price
        self.lam = lam
        # θ = [b, a]ᵀ  (intercept-first convention)
        self._theta = np.array([5.0, 0.5])
        # P = covariance matrix; large diagonal = high initial uncertainty
        self._P = np.eye(2) * 1000.0
        self.u_L_hist = []
        self.u_F_hist = []

    @property
    def a(self):
        return float(self._theta[1])

    @property
    def b(self):
        return float(self._theta[0])

    def _batch_init_rls(self):
        """Warm-start θ and P from batch OLS on loaded historical data."""
        if len(self.u_L_hist) < 2:
            return
        X = np.column_stack([np.ones(len(self.u_L_hist)), self.u_L_hist])
        Y = np.array(self.u_F_hist)
        XtX = X.T @ X
        if abs(np.linalg.det(XtX)) > 1e-10:
            self._theta = np.linalg.solve(XtX, X.T @ Y)
            self._P = np.linalg.inv(XtX)

    def _rls_update(self, u_L_new, u_F_new):
        """One recursive update step (Lecture 6, equations 6.1-6.3)."""
        phi = np.array([1.0, u_L_new])
        denom = self.lam + phi @ self._P @ phi
        K = (self._P @ phi) / denom
        self._theta = self._theta + K * (u_F_new - phi @ self._theta)
        self._P = (self._P - np.outer(K, phi @ self._P)) / self.lam

    def _record_date(self, t):
        """Fetch prices for date t and store in history list."""
        try:
            prices = self.get_price_from_date(t)
            if prices is None:
                return
            lp, fp = float(prices[0]), float(prices[1])
            self.u_L_hist.append(lp)
            self.u_F_hist.append(fp)
        except Exception:
            pass

    def _record_and_update(self, t):
        """Fetch day-t prices and immediately run one RLS step."""
        try:
            prices = self.get_price_from_date(t)
            if prices is None:
                return
            lp, fp = float(prices[0]), float(prices[1])
            self._rls_update(lp, fp)
        except Exception:
            pass

    def _optimal_price(self):
        A = 100.0 + 3.0 * self.b
        B = 3.0 * self.a - 5.0
        if B >= 0:
            p_max = self._expected_profit(self.max_price)
            p_min = self._expected_profit(self.min_price)
            return self.max_price if p_max >= p_min else self.min_price
        u_opt = (B - A) / (2.0 * B)
        return float(max(self.min_price, min(self.max_price, u_opt)))

    def _expected_profit(self, u_L):
        u_F = self.a * u_L + self.b
        return (u_L - 1.0) * (100.0 - 5.0 * u_L + 3.0 * u_F)

    def end_simulation(self):
        self.u_L_hist.clear()
        self.u_F_hist.clear()
        self._theta = np.array([5.0, 0.5])
        self._P = np.eye(2) * 1000.0

class _EWLSMixin:
    """EWLS + Stackelberg pricing logic. Must be combined with a Leader subclass."""

    def _init_ewls(self, min_price=1.0, max_price=100.0, decay=0.97):
        self.min_price = min_price
        self.max_price = max_price
        self.decay = decay
        self.a = 0.5
        self.b = 5.0
        self.u_L_hist = []
        self.u_F_hist = []

    def _record_date(self, t):
        """Fetch prices for date t and append only if both are valid numbers."""
        try:
            prices = self.get_price_from_date(t)
            if prices is None:
                return
            lp, fp = prices[0], prices[1]
            if isinstance(lp, (int, float)) and isinstance(fp, (int, float)):
                self.u_L_hist.append(float(lp))
                self.u_F_hist.append(float(fp))
        except Exception:
            pass

    def predict_follower_price(self, u_L: float) -> float:
        return self.a * u_L + self.b

    def compute_profit(self, u_L: float, u_F: float) -> float:
        S_L = 100.0 - 5.0 * u_L + 3.0 * u_F
        return (u_L - 1.0) * S_L

    def expected_profit(self, u_L: float) -> float:
        return self.compute_profit(u_L, self.predict_follower_price(u_L))

    def _filter_outliers(self, xs, ys):
        n = len(ys)
        if n < 10:
            return xs, ys
        sorted_y = sorted(ys)
        q1 = sorted_y[n // 4]
        q3 = sorted_y[(3 * n) // 4]
        threshold = q3 + 3.0 * (q3 - q1)
        xs_f = [x for x, y in zip(xs, ys) if y <= threshold]
        ys_f = [y for x, y in zip(xs, ys) if y <= threshold]
        return xs_f, ys_f

    def _fit_ewls(self):
        xs, ys = self._filter_outliers(self.u_L_hist, self.u_F_hist)
        n = len(xs)
        if n < 2:
            return
        weights = [self.decay ** (n - 1 - i) for i in range(n)]
        sw   = sum(weights)
        swx  = sum(w * x     for w, x    in zip(weights, xs))
        swy  = sum(w * y     for w, y    in zip(weights, ys))
        swxx = sum(w * x * x for w, x    in zip(weights, xs))
        swxy = sum(w * x * y for w, x, y in zip(weights, xs, ys))
        denom = sw * swxx - swx ** 2
        if abs(denom) < 1e-10:
            self.b = swy / sw
        else:
            self.a = (sw * swxy - swx * swy) / denom
            self.b = (swy - self.a * swx) / sw

    def _optimal_price(self):
        A = 100.0 + 3.0 * self.b
        B = 3.0 * self.a - 5.0
        if B >= 0:
            p_max = self.expected_profit(self.max_price)
            p_min = self.expected_profit(self.min_price)
            return self.max_price if p_max >= p_min else self.min_price
        u_opt = (B - A) / (2.0 * B)
        return max(self.min_price, min(self.max_price, u_opt))

    def start_simulation(self):
        for t in range(1, 101):
            self._record_date(t)
        self._fit_ewls()

    def new_price(self, date: int) -> float:
        if date > 101:
            self._record_date(date - 1)
            self._fit_ewls()
        return self._optimal_price()

    def end_simulation(self):
        self.u_L_hist.clear()
        self.u_F_hist.clear()
        self.a = 0.5
        self.b = 5.0


# -----------------------------------------------------------------------------
# LeaderMK1 — direct Leader subclass (appears in GUI dropdown)
# Explore 5→15→30 first: follower has only seen prices ~1.8 historically so
# jumping straight to the fitted optimum (~20) crashes the engine.
# -----------------------------------------------------------------------------
class LeaderMK1(Leader, _EWLSMixin):
    _EXPLORE_PRICES = {101: 5.0, 102: 15.0, 103: 30.0}

    def __init__(self, name, engine):
        super().__init__(name, engine)
        self._init_ewls(min_price=1.0, max_price=50.0, decay=0.95)

    def new_price(self, date: int) -> float:
        if date > 101:
            self._record_date(date - 1)
            self._fit_ewls()
        if date in self._EXPLORE_PRICES:
            return self._EXPLORE_PRICES[date]
        return self._optimal_price()


# -----------------------------------------------------------------------------
# LeaderMK2 — direct Leader subclass (appears in GUI dropdown)
# Loads only days 51-100 (pre-day-50 data is a different regime).
# Explore 3→8→15 first before going greedy — same engine safety concern,
# and the slope estimate from narrow historical range needs new data.
# -----------------------------------------------------------------------------
class LeaderMK2(Leader, _EWLSMixin):
    _EXPLORE_PRICES = {101: 3.0, 102: 8.0, 103: 15.0}

    def __init__(self, name, engine):
        super().__init__(name, engine)
        self._init_ewls(min_price=1.0, max_price=50.0, decay=0.97)

    def start_simulation(self):
        for t in range(51, 101):
            self._record_date(t)
        self._fit_ewls()

    def new_price(self, date: int) -> float:
        if date > 101:
            self._record_date(date - 1)
            self._fit_ewls()
        if date in self._EXPLORE_PRICES:
            return self._EXPLORE_PRICES[date]
        return self._optimal_price()


# -----------------------------------------------------------------------------
# LeaderMK3 — direct Leader subclass (appears in GUI dropdown)
# Follower price is near-constant (~1.34); forces a=0, fits intercept only.
# Loads last 30 days; max_price=15 (game rule). Explore 3→6→10 first.
# -----------------------------------------------------------------------------
class LeaderMK3(Leader, _EWLSMixin):
    _EXPLORE_PRICES = {101: 3.0, 102: 6.0, 103: 10.0}

    def __init__(self, name, engine):
        super().__init__(name, engine)
        self._init_ewls(min_price=1.0, max_price=15.0, decay=0.93)

    def start_simulation(self):
        for t in range(71, 101):
            self._record_date(t)
        self._fit_ewls()

    def new_price(self, date: int) -> float:
        if date > 101:
            self._record_date(date - 1)
            self._fit_ewls()
        if date in self._EXPLORE_PRICES:
            return self._EXPLORE_PRICES[date]
        return self._optimal_price()

    def _fit_ewls(self):
        xs, ys = self._filter_outliers(self.u_L_hist, self.u_F_hist)
        n = len(ys)
        if n < 1:
            return
        weights = [self.decay ** (n - 1 - i) for i in range(n)]
        sw  = sum(weights)
        swy = sum(w * y for w, y in zip(weights, ys))
        self.a = 0.0
        self.b = swy / sw
```

```python
# =============================================================================
# LeaderMK1 — plays MK1 and the hidden follower MK4
# =============================================================================
# Strategy summary (RLS edition, Lecture 6)
# -----------------------------------------
# Phase 0 (start_simulation): Load 100 days of historical data and warm-start
#                             RLS theta + P from a single batch OLS solve.
# Phase 1 (day 101 only):     Single explore at u_L = 19 — one wide-range
#                             datapoint is enough for RLS to pin down theta
#                             via its exact recursive Bayesian update.
# Phase 2 (days 102-130):     Greedy Stackelberg optimum from the live RLS
#                             estimate, protected by a break-even safety guard
#                             AND a hard ceiling at u_L = 20 (the price above
#                             which demand can become negative under worst-
#                             case follower behavior).
#
# Why these choices (Member 1 priorities, all addressed)
# ------------------------------------------------------
#   #1 Catastrophic explore days eliminated. The previous 3-day explore
#      {5, 15, 30} crashed S_L to ~-135 on day 103, costing ~£1,150 in one
#      day. RLS converges to good theta from a *single* well-placed update,
#      so we now do exactly one explore day at u_L = 19 (just below the
#      worst-case break-even ceiling of 20). This is the largest gain.
#   #2 Wider-but-safer explore. Under RLS, a single datapoint at u_L = 19
#      gives a higher information content than three datapoints clustered
#      near 1.8 (that's what the historical batch already provides). Sweep
#      tested 1-day, 2-day and 3-day explore schedules — single-day won.
#   #3 max_price LOWERED to 20, not raised. With max=100, when RLS briefly
#      fits a slope a >= 5/3 (slope >= 1.667), `_optimal_price()` returns
#      max_price as the optimum — a guaranteed catastrophe at u_L = 100.
#      Capping at 20 makes the worst case S_L = 100 - 100 + 3*u_F >= 0
#      under any non-negative follower price. Sweep showed this is worth
#      ~£1,400/run vs max=100.
#   #4 lambda set to 0.97. RLS converges so quickly after the explore day
#      that lambda barely matters — the entire range 0.90 to 1.00 differed
#      by only ~£70 in tests. 0.97 is a balanced choice that still tracks
#      slow drift in the follower's reaction.
#   #5 Zero hard-coded follower numbers. The explore price 19, the cap 20,
#      and the safety-guard formulae are all derived from the demand model
#      (S_L = 100 - 5*u_L + 3*u_F, fixed for all 130 days) and the live
#      RLS estimate. No MK1-specific values appear, so the strategy
#      generalises to MK4 by construction.
#
# What is NOT changed (per Member 1 instructions)
# -----------------------------------------------
#   - _RLSMixin class itself
#   - The recursive update equations (Lecture 6 6.1-6.3)
#   - The 100-day history ingestion is preserved, just done explicitly here
#     because _RLSMixin (unlike _EWLSMixin) does not provide a default
#     start_simulation.
# =============================================================================


class Leader_MK1(Leader, _RLSMixin):
    """Leader designed to play against MK1 and the hidden follower MK4.

    Inherits directly from `Leader` so it appears in the GUI dropdown
    (Leader.__subclasses__() only returns direct subclasses), and from
    `_RLSMixin` for the recursive least-squares parameter estimator
    plus the closed-form Stackelberg optimum solver.
    """

    # -------------------------------------------------------------------------
    # Explore schedule (single day)
    # -------------------------------------------------------------------------
    # Just one explore day at u_L = 19. This is the "safe ceiling": the
    # theoretical break-even price for the demand model is 20 when u_F = 0
    # (worst case follower), and the historical reaction-function fit gives
    # an interior optimum of ~19.6, so 19 is both informative and profitable.
    #
    # Why not 3 days like before? RLS is a Bayesian update — one well-placed
    # datapoint at u_L = 19, contrasted with the 100 historical points at
    # u_L ~ 1.8, gives an enormous information gain in a single step. Adding
    # more explore days costs greedy-day profit without proportional gain;
    # sweep tested 1/2/3-day schedules and 1-day won by ~£800 per run.
    _EXPLORE_PRICES = {101: 19.0}

    # Predicted demand must stay above this margin or the safety guard kicks in.
    _DEMAND_MARGIN = 1.0

    # Cushion factor applied below the projected break-even price.
    _SAFE_FRACTION = 0.95

    def __init__(self, name, engine):
        super().__init__(name, engine)
        # max_price = 20 is the principled hard cap derived from the demand
        # model: at u_L = 20, S_L = 100 - 100 + 3*u_F = 3*u_F, which is
        # non-negative for any non-negative follower price. So even if RLS
        # fits a degenerate slope (a >= 5/3) and `_optimal_price` returns
        # max_price, we cannot crash through into negative-demand territory.
        # lam = 0.97 -> mild forgetting; RLS converges fast, so lambda is
        # not very sensitive. Sweep range 0.90-1.00 differed by only ~£70.
        self._init_rls(min_price=1.0, max_price=20.0, lam=0.97)

    # -------------------------------------------------------------------------
    # Helper: ingest a day's data AND apply one RLS step
    # -------------------------------------------------------------------------
    # _RLSMixin provides _record_date (history-only) and _record_and_update
    # (RLS-only). For the live phase we want BOTH: recursive theta updates
    # plus a tracked history (the safety guard's fallback uses u_L_hist).
    def _ingest_and_update(self, t):
        try:
            prices = self.get_price_from_date(t)
            if prices is None:
                return
            lp, fp = float(prices[0]), float(prices[1])
            self.u_L_hist.append(lp)
            self.u_F_hist.append(fp)
            self._rls_update(lp, fp)
        except Exception:
            pass

    # -------------------------------------------------------------------------
    # start_simulation must be defined — _RLSMixin does not provide one
    # -------------------------------------------------------------------------
    def start_simulation(self):
        # Stage 1: ingest 100 days of historical data into u_L_hist/u_F_hist.
        for t in range(1, 101):
            self._record_date(t)
        # Stage 2: warm-start theta from batch OLS (sets P = inv(X^T X) too).
        # This gives a strong starting point that the explore day will refine.
        self._batch_init_rls()

    # -------------------------------------------------------------------------
    # Safety guard — derived from the (known, fixed) demand model
    # -------------------------------------------------------------------------
    def _safe_optimal_price(self) -> float:
        """Wrap `_optimal_price` with a break-even safety guard.

        Uses the live RLS coefficients (a, b) and the *known* demand model
        S_L = 100 - 5*u_L + 3*u_F to project demand at the proposed price.
        If projected demand falls below the configured margin, fall back to
        the highest price consistent with that margin under the linear fit:

            100 - 5*u + 3*(a*u + b) = margin
            =>  u_safe = (100 + 3*b - margin) / (5 - 3*a)        when a < 5/3

        Note: with max_price = 20, the linear-extrapolation guard is rarely
        triggered in practice (the cap usually does the work first), but it
        adds belt-and-braces safety against edge cases.
        """
        raw = self._optimal_price()

        # Predict demand at the proposed price using the current RLS fit.
        pred_uF = self.a * raw + self.b
        pred_SL = 100.0 - 5.0 * raw + 3.0 * pred_uF

        if pred_SL >= self._DEMAND_MARGIN:
            return raw

        # Demand projection is too low — recompute a safe ceiling.
        denom = 5.0 - 3.0 * self.a
        if denom > 1e-6:                                  # i.e. a < 5/3
            safe = (100.0 + 3.0 * self.b - self._DEMAND_MARGIN) / denom
            safe *= self._SAFE_FRACTION                   # extra cushion
            return max(self.min_price, min(safe, self.max_price))

        # Degenerate fit (slope >= 5/3): the linear extrapolation is not
        # trustworthy. Stay close to the highest price we have actually
        # observed and trusted, capped at max_price.
        if self.u_L_hist:
            return max(self.min_price, min(self.max_price, max(self.u_L_hist)))
        return self.min_price

    # -------------------------------------------------------------------------
    # Daily price decision
    # -------------------------------------------------------------------------
    def new_price(self, date: int) -> float:
        # On every greedy day, ingest yesterday's outcome and update RLS.
        if date > 101:
            self._ingest_and_update(date - 1)

        # Single explore day at u_L = 19 to anchor the slope estimate.
        if date in self._EXPLORE_PRICES:
            return self._EXPLORE_PRICES[date]

        # Greedy phase: live optimum, defended by the safety guard and the
        # hard max_price cap.
        return self._safe_optimal_price()
```
